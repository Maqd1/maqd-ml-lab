## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

csv_url = "https://raw.githubusercontent.com/Maqd1/maqd-ml-lab/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(csv_url)

print("Dataset shape:", df.shape)
print("Number of pages:", len(df))
print("Number of columns:", len(df.columns))
print("Available columns:")
print(df.columns.tolist())

visible_pages = df[df["impressions_90d"] >= 500]

print("Pages with at least 500 impressions:", len(visible_pages))

high_visibility_low_ctr = df[
    (df["impressions_90d"] >= 500) &
    (df["ctr"] < 1)
]

print("Pages with high impressions and CTR below 1%:", len(high_visibility_low_ctr))

stale_visible_pages = df[
    (df["impressions_90d"] >= 500) &
    (df["days_since_last_update"] >= 180)
]

print("Visible pages not updated for at least 180 days:",
      len(stale_visible_pages))

print(df["position_tier"].value_counts(dropna=False))

position_groups = df.groupby("position_tier")

ctr_by_position = position_groups["ctr"].mean()

print(ctr_by_position)

print(df["content_type"].value_counts(dropna=False))

engagement_by_content_type = df.groupby("content_type")["engagement_rate"].mean()

print(engagement_by_content_type)

df["sessions_90d"].describe()
```
