## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

### Quick look at the data

The starter dataset contains **30,000 content pages and 44 columns**. Of these, **16,726 pages have at least 500 impressions** over the 90-day period. This gives us a substantial group of pages with measurable search visibility.

An initial exploratory filter found **16,167 pages with at least 500 impressions and a CTR below 1%**. This suggests that many visible pages receive relatively few clicks compared with their impressions. However, a low CTR alone does not prove that a page is an improvement opportunity. Search position and other contextual factors must also be considered.

The dataset also shows that session volume is uneven. The median page has only **7 sessions**, while **8,073 pages have between 1 and 2 sessions**. This matters because engagement measurements based on very few sessions provide weaker evidence than measurements based on larger samples.

The average CTR differs across search-position groups, and average engagement differs across content types. Therefore, pages should be compared with relevant groups rather than judged using one universal performance threshold.

These initial findings support an opportunity-led approach: identify pages with possible improvement potential, consider the context and strength of their data, and prioritise them for human review rather than automatically labelling them as pages that need refreshing.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

csv_url = "https://raw.githubusercontent.com/Maqd1/maqd-ml-lab/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(csv_url)

# 1. Dataset size
print("Total pages:", len(df))
print("Total columns:", len(df.columns))

# 2. Pages with measurable visibility
visible_pages = df[df["impressions_90d"] >= 500]

print(
    "Pages with at least 500 impressions:",
    len(visible_pages)
)

# 3. Visible pages with relatively low CTR
high_visibility_low_ctr = df[
    (df["impressions_90d"] >= 500) &
    (df["ctr"] < 1)
]

print(
    "Visible pages with CTR below 1%:",
    len(high_visibility_low_ctr)
)

# 4. Pages with very limited session data
low_session_pages = df[df["sessions_90d"] <= 2]

print(
    "Pages with only 1–2 sessions:",
    len(low_session_pages)
)
```
